```python
import pandas as pd 
import glob
import os
```

```python
path = os.getcwd()
```

```python
os.listdir()
```

```python
path = os.path.join(path,'Data')
```

```python
all_files = glob.glob(os.path.join(path,"reviewers_data_*.csv"))
```

```python
all_files
```

```python
my_list = []
for f in all_files:
    df = pd.read_csv(f,index_col=None)
    my_list.append(df)
res = pd.concat(my_list,axis=0,ignore_index=True)
```

```python
res.head()
```

```python
print(res[res['book_id'] == 'book_id'])
```

```python
res.columns
```

```python
res.duplicated(subset=['book_id','user_id']).sum()
```

```python
temp = res.drop_duplicates(subset=['book_id','user_id'],keep='last')
```

```python
temp = temp.drop(columns=['user_name','comment'])
```

```python
temp.columns
```

```python
print(f"Before drop duplicates : {len(res)}\nAfter drop duplicates : {len(temp)}")
```

```python
temp.isnull().sum()
```

```python
temp['rating'].describe()
```

```python
temp2 = temp[temp['rating']>0]
```

```python
temp2['rating'].min()
```

```python
user_counts = temp2['user_id'].value_counts()
user_counts
```

```python
temp2 = temp2[temp2['user_id'].isin(user_counts[user_counts>5].index)]
```

```python
print(f"Before filter : {len(temp)}\nAfter filter : {len(temp2)}")
```

```python
# maximum sparsity requirement <= 99%
user_counts = temp2['user_id'].value_counts()
book_counts = temp2['book_id'].value_counts()
sparsity =( 1 - (len(temp2)/(len(user_counts)*len(book_counts))))*100
if sparsity <= 99 :
    print(f"The sparsity is {sparsity:.2f}%")
else:
    print("not yet achieved under 99%")
```

```python
path = os.path.join("Data",'Clean_reviewers_data_without_comment.csv')
temp2.to_csv(path,index=False)
```
