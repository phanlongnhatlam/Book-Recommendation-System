```python
import glob
import os
import pandas as pd
import matplotlib as plt
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = os.getcwd()
```

```python
path = os.path.join(path,'Data')
```

```python
all_files = glob.glob(os.path.join(path,"books_data_*.csv"))
```

```python
all_files
```

```python
my_list = []
for f in all_files:
    df = pd.read_csv(f,index_col=None)
    my_list.append(df)
df_books = pd.concat(my_list,axis=0,ignore_index=True)
```

```python
df_books1 = df_books.dropna(subset=['title']).copy()
display(df_books1.isnull().sum())
```

```python
print(len(df_books1))
df_books1 = df_books1.drop_duplicates(subset='work_id')
print(len(df_books1))
```

```python
print(len(df_books1))
df_books1 = df_books1.dropna(subset='work_id')
print(len(df_books1))
```

```python
df = df_books1.drop(columns='work_id')
```

```python
df.head(10)
```

```python
df_books = df
```

```python
df_books['content_soup'] = (
    (df_books['author']+ " ")*3  +
    (df_books['genres']+ " ")*3  +
    df_books['description']
)
```

```python
df_books.head(10)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df_books['avg_rating'], bins=20, kde=True, color='coral')
plt.title('Insight : Phân bố Điểm đánh giá', fontsize=14, fontweight='bold')
plt.xlabel('Điểm số (Rating)', fontsize=12)
plt.ylabel('Số lượng sách', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
#top 10 tác giả xuất hiện nhiều nhất
top_authors = df_books['author'].value_counts().head(10)

sns.barplot(x=top_authors.values, y=top_authors.index, palette='Blues_r', hue=top_authors.index, legend=False)
plt.title('Insight : Top 10 Tác giả có số lượng sách áp đảo nhất', fontsize=14, fontweight='bold')
plt.xlabel('Số lượng sách', fontsize=12)
plt.ylabel('Tên tác giả', fontsize=12)
plt.tight_layout()
plt.show()
```
